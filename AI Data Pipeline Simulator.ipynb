```python
import numpy as np

# 50 samples, 4 features (age, income, score, usage_time)
data = np.random.randint(50, 150, (50, 4))

# Compute feature-wise mean and standard deviation
mean = np.mean(data, axis=0)
std = np.std(data, axis=0)

# Standardization: (x - mean) / std
norm = (data - mean) / std

print("Means:", np.round(mean, 2))
print("Standard deviations:", np.round(std, 2))
print("First 3 rows (normalized):\n", np.round(norm[:3], 2))
```
